
### 🧠 Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) is the process of analyzing datasets to summarize their main characteristics, often using visual methods. It helps in understanding the data structure and detecting anomalies, patterns, or relationships between variables.

---

### 🎯 Objective of EDA

- To understand the structure and distribution of the data.
- To clean and prepare the data for modeling.
- To identify relationships between variables.
- To reduce data redundancy and improve data quality.

---

### 🪜 Common EDA Steps

1. Understand the data.
2. Clean the data.
3. Analyze relationships between variables.
4. Handle missing values and outliers.
5. Perform variable transformations and feature engineering.

---

### 🧹 Data Cleaning Process

1. **Variable Identification**  
   - **Independent Variables**: Variables that influence the target. These are inputs or predictors.
   - **Dependent Variable**: The outcome or response variable that depends on the independent variables.

2. **Univariate Analysis**  
   - Analyze one variable at a time using histograms, boxplots, etc.

3. **Bivariate Analysis**  
   - Explore relationships between two variables using scatterplots, correlation, etc.

4. **Handling Missing Values**  
   - Impute with mean, median, mode, or advanced methods like KNN.

5. **Outlier Treatment**  
   - Detect and treat outliers using boxplots, Z-score, or IQR.

6. **Variable Transformation**  
   - Normalize or scale features if needed.

7. **Feature Creation**  
   - Engineer new variables that help improve model performance.
